Purpose: Use previously generated BED files for 1kb promoters of all genes in each parental species (*Y. aloifolia* & *Y. filamentosa*) to generate a promoter BED file matching the *in silico Y. gloriosa* chromosome names.<br>

Date initiated: Mar. 5, 2026

```python
import pandas as pd
```

```python
yabed = pd.read_csv("/home/leviathan22/yucca-genomics/Ya_1kb_promoters.bed",sep="\t",header=None)
yfbed = pd.read_csv("/home/leviathan22/yucca-genomics/Yf_1kb_promoters.bed",sep="\t",header=None)
```

```python
yabed.head()
```

```python
def correct_chroms(bed,sp):
    d = {}
    for i in bed[0].unique():
        d[i] = sp+"_"+i
    bed[0] = bed[0].map(d)
    return bed
```

```python
yabed = correct_chroms(yabed,"Ya")
```

```python
yfbed = correct_chroms(yfbed,"Yf")
```

```python
ygbed = pd.concat([yabed,yfbed])
```

```python
ygbed.to_csv("/home/leviathan22/yucca-genomics/Yg_1kb_promoters.bed",sep="\t",header=False,index=False)
```
