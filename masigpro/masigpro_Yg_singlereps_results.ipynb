```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
res = pd.read_csv("./hck6_Yg_singlereps.tsv",sep="\t",header="infer")
res.head()
```

```python
# how many total unique time-structured genes?
len(res["GeneID"].unique())
```

```python
len(res.index)
```

```python
# add subgenome column
sg = []
for i in res["GeneID"]:
    if i.startswith("Yucal"):
        sg.append("Ya")
    else:
        sg.append("Yf")
```

```python
res["subgenome"] = sg
res.head()
```

```python
len(res[res["subgenome"]=="Ya"].index)
```

```python
len(res[res["subgenome"]=="Yf"].index)
```
